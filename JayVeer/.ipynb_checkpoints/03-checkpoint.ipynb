```python
a = 25
b = 5
print("Addition of a & b:",a+b,"| Substraction of a & b:",a-b)

print("Multiplication of a & b:",a*b,"| Division of a & b:",a/b)

print("Square of a & b:",a**b,"| Floor division of a & b:",a//b)

print("Modulo of a & b:",a%b)
```

```python
print(10 % 2, 11 % 2)
print(1779 % 10, 1234 // 10)
print(70 % 7 == 0)
print(2 ** 10, 2**5, sep="|")
print(9 ** 0.5)   
print(10 ** -22)
print(-3 ** 2) 
print(--3 ** 2)  #or
print((-3) ** 2)
```

```python
print("absolute:",abs(-4.5))
print("Round figure:",round(3.7), "| decimal shiffting by 2 digit",round(3.14159, 3))                          
print("smallest number:",min(4, 9, 2), "| Biggest number:", max(4, 9, 2))
print(divmod(17, 5))
```

```python
import math
print(math.sqrt(144))
print(math.pi)
print(math.floor(2.7), math.ceil(2.1))   # round down / round up
print(math.hypot(3, 4))                  # length of the hypotenuse
print(math.isclose(0.1 + 0.2, 0.4))
```
